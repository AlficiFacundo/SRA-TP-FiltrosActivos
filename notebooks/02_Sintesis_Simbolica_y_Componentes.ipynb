```python
import sympy as sp
import numpy as np
import pandas as pd
s=sp.symbols('s')
Vi,Vo=sp.symbols('Vi Vo')
R,C=sp.symbols('R C',positive=True,real=True)
```

### Ejemplo guiado: red RC pasabajos descargada
Aplicar KCL: $(V_i-V_o)/R=sCV_o$. Obtener la transferencia como cociente de variables simbólicas.

```python
eq=sp.Eq((Vi-Vo)/R,s*C*Vo)
H=sp.factor(sp.solve(eq,Vo)[0]/Vi)
print('H_RC(s)=',H)
```
